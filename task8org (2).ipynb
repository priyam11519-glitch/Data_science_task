```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("C:/Users/ELCOT/Downloads/hotel_bookings.csv")
```

```python
df.shape
```

```python
df.columns
```

```python
df.dtypes
```

```python
df.isnull().sum()
```

```python
for col in df.select_dtypes(include='number').columns:
    df[col] = df[col].fillna(df[col].median())
```

```python
for col in df.select_dtypes(include='str').columns:
    df[col] = df[col].fillna(df[col].mode()[0])
```

```python
df.isnull().sum()
```

```python
num = ['lead_time', 'adr', 'stays_in_week_nights', 'total_of_special_requests']
```

```python
for col in num:
    plt.figure(figsize=(6,4))
    plt.hist(df[col], bins=30)
    plt.title(col)
    plt.xlabel(col)
    plt.ylabel('Frequency')
    plt.show()
    
```

```python
for col in num:
    plt.figure(figsize=(6,4))
    sns.boxplot(x=df[col])
    plt.title(col)
    plt.show()
```

```python
for col in num:
    s = df[col].skew()
    if s > 0:
        print(col, "Positive Skew")
    elif s < 0:
        print(col, "Negative Skew")
    else:
        print(col, "Normal Distribution")
```
