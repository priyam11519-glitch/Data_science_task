```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("C:/Users/ELCOT/Downloads/hotel_bookings (1).csv")
```

```python
print(df.shape)
print(df.columns)
print(df.dtypes)
print(df.isnull().sum())
```

```python
df = df.dropna()
print(df.isnull().sum())
```

```python
sns.scatterplot(x='lead_time', y='adr', data=df)
plt.title('Lead Time vs ADR')
plt.show()
```

```python
sns.violinplot(x='hotel', y='adr', data=df)
plt.title('Hotel Type vs ADR')
plt.show()
```

```python
sns.boxplot(x='lead_time', data=df)
plt.title('Box Plot of Lead Time')
plt.show()

print("Skewness:", df['lead_time'].skew())
```

```python
sns.barplot(x='hotel', y='adr', data=df)
plt.title('Average ADR by Hotel Type')
plt.show()
```
